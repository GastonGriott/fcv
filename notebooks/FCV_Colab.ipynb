## 4. Corrida justa (in-kernel, paralelo)

```
import importlib, fcv.core, fcv.methods, fcv.run_parallel
for m in (fcv.core, fcv.methods, fcv.run_parallel): importlib.reload(m)
from fcv import run_parallel
# grilla paper-grade (8 nucleos): mapea la frontera floor-fail por familia
GRID={'knapsack':[30,40,50,60,80], 'scp':[40,60,90,120], 'uflp':[12,16,18]}
run_parallel.run(jobs=-1, grid=GRID, instances=5, n_seeds=30,
                 methods_list=('floor','ils','ga'),
                 out_csv='fcv_grid_paper.csv')
```
